```python
no_hours = len(df.axes[0])//60

ls_h=[]
for hour_cnt in tqdm(range(no_hours)):

  df_temp =df[hour_cnt*60:(hour_cnt+1)*60]

  date, time =df_temp['0'].tolist()[0].split()

  hh, mm, ss =time.split(':')

  hh_ =str('{:02d}'.format(int(hh)+1))

  time_new_ls =[hh_, mm, ss]
  time_new =(':').join(time_new_ls)

  time_range = time+'-'+time_new
  date_cum_time = date+' '+time_range

  di_h={}
  di_h['date_cum_time_duration'] =date_cum_time

  for colm in df_temp.columns:
    if(colm =='0'): continue

    di_h[colm] = df_temp[colm].sum(skipna=True)

  ls_h.append(di_h)

df_h =pd.DataFrame(ls_h)
```

# Observation:-

1.   Although its a general notion that the power consumption tends to increase with as the outside temperture increases (given that AC is in operation for same durations at varying outside temperature), just because the compressor has to work more with increaseing temperature differnece between source and sink, our conclusions don't suggest the same.
2.  The line subplots qualitatively show that outside average daily temperture is not positively correlated to the average daily power consumption of each AC. This fact is further confirmed by the quantitaive correlation scores as depicated in above heatmap.

3.   Although many of the AC/temperature correlation are positive with quite a lot having a score near 0.50 (AC5, AC6, AC7, AC8, AC11, AC 12. AC 14, AC17); none have significant high score (> 0.50, maybe atleast 0.7) to ascertain that there is high positive correlation between AC power consumption and outside temperature. Infact there are cases like of AC 18 where the correlation score is infact negative (-0.24). These evidence suggests that in relation with the varying usage of different ACs, it all boils to residents of rooms corresponding to AC using the ACs in varying capacities according to their personal comfort, habits and adaptiblity that may not significantly change with change in weather. For instance, the residents of room 18 (AC 18) have used that AC at maximum load at most times whereas residents of room 5 (AC 5) have used that AC at lower loads at most times.

**preprocessing**

```python
ls_preprocess =[]
for idx in tqdm(range(len(df_new.axes[0]))):
  d_row =df_new.iloc[idx].to_dict()

  d_cyclic={}
  date_time =d_row['date_cum_time_duration']
  date, timespan = date_time.split()

  lower_time, upper_time = timespan.split('-')
  lower_hour = int(lower_time.split(':')[0])

  d_cyclic['date_time'] = date+' '+lower_time
  d_cyclic['lower_hour_bound'] =lower_hour

  for col in df_new.columns:
    if(col =='date_cum_time_duration'): continue

    d_cyclic[col] = d_row[col]

  ls_preprocess.append(d_cyclic)

df_preprocess = pd.DataFrame(ls_preprocess)

# Creating the cyclical daily feature
df_preprocess["day_cos"] = [np.cos(x * (2 * np.pi / 24)) for x in df_preprocess["lower_hour_bound"]]
df_preprocess["day_sin"] = [np.sin(x * (2 * np.pi / 24)) for x in df_preprocess["lower_hour_bound"]]

# Creating the cyclical monthly feature
df_preprocess['timestamp'] =[datetime.timestamp(datetime.strptime(x, "%Y-%m-%d %H:%M:%S")) for x in df_preprocess["date_time"]]

# Seconds in day
s = 24 * 60 * 60
# Seconds in year
year = (365.25) * s

df_preprocess["month_cos"] = [np.cos((x) * (2 * np.pi / year)) for x in df_preprocess["timestamp"]]
df_preprocess["month_sin"] = [np.sin((x) * (2 * np.pi / year)) for x in df_preprocess["timestamp"]]
```

```python
class process:

  def __init__(self, lag, n_ahead, ac_name, df, min_delta, patience, epoch, batch_size, train_days,  lr, hidden_ouput_dim, feature_column):
    self.lag =lag
    self.n_ahead =n_ahead
    self.ac_name =ac_name
    self.train_days =train_days

    self.feature_column= feature_column

    self.df =df
    # self.dfnew = self.df[['date_time', ac_name, 'day_cos', 'day_sin', 'month_cos', 'month_sin']]
    # self.dfnew = self.df[['date_time', ac_name]]
    self.dfnew = self.df[self.feature_column]

    self.epoch =epoch
    self.batch_size =batch_size
    self.lr =lr
    self.min_delta =min_delta# successeive iterations must have absolute diff of mointored metric(here validation loss) greater than this to count as improved score
    self.patience = patience# minm iterations to look for mointored metric(here validation loss) improvemnet
    self.hidden_ouput_dim = hidden_ouput_dim

  def data_split_cum_preparation(self):

    self.df_train = self.dfnew.iloc[:self.train_days*24,1:]# discard the date_time colums(given ac hourly power consumption becomes 0th column)
    self.df_val = self.dfnew.iloc[self.train_days*24:,1:]# discard the date_time colums(given ac hourly power consumption becomes 0th column)

    self.n_train, self.n_val = len(self.df_train.axes[0]), len(self.df_val.axes[0])
    print(self.n_train, self.n_val)
    self.minmax_new =MinMaxScaler(feature_range=(0,1))

    self.minmax_new.fit(self.df_train)

    self.train_arr = self.minmax_new.transform(self.df_train)
    self.val_arr =self.minmax_new.transform(self.df_val)

    self.X_train, self.Y_train = [],[]

    for i in range(self.n_train-self.lag- self.n_ahead+1):
      self.X_train.append(self.train_arr[i:(i+self.lag), :])# access elements from rows i to i+lag(with all clomns included)
      self.Y_train.append(self.train_arr[(i+self.lag):(i+self.lag+self.n_ahead), 0])# hourly power consumption of given ac for n_ahead hrs

    # for i in range(self.lag, self.n_train):
    #   self.X_train.append(self.train_arr[i:(i-self.lag), :])
    #   self.Y_train.append(self.train_arr[i, 0])

    self.X_val, self.Y_val =[],[]

    for j in range(self.n_val-self.lag- self.n_ahead+1):
      self.X_val.append(self.val_arr[j:(j+self.lag),:])# access elements from rows i to i+lag(with all clomns included)
      self.Y_val.append(self.val_arr[(j+self.lag):(j+self.lag+self.n_ahead),0])# hourly power consumption of given ac for n_ahead hrs
    # for j in range(self.lag, self.n_val):
    #   self.X_val.append(self.val_arr[j:(j-self.lag),:])
    #   self.Y_val.append(self.val_arr[j,0])


    self.X_train, self.Y_train = np.array(self.X_train), np.array(self.Y_train)
    self.X_val, self.Y_val = np.array(self.X_val), np.array(self.Y_val)

  def arctitecture(self):

    self.model =Sequential()
    #Adding the first LSTM layer and some Dropout regularisation
    self.model.add(LSTM(units = self.hidden_ouput_dim, return_sequences = True, input_shape = (self.X_train.shape[1], self.X_train.shape[2])))
    self.model.add(Dropout(0.2))

    # Adding a second LSTM layer and some Dropout regularisation
    self.model.add(LSTM(units = self.hidden_ouput_dim, return_sequences = True))
    self.model.add(Dropout(0.2))

    # # Adding a third LSTM layer and some Dropout regularisation
    # self.model.add(LSTM(units = self.hidden_ouput_dim, return_sequences = True))
    # self.model.add(Dropout(0.2))

    # Adding the output layer
    self.model.add(Dense(units = self.n_ahead))

  def trainCallback(self):
    return EarlyStopping(monitor='val_loss', patience=self.patience, min_delta=self.min_delta)


  def training(self):
    # Initiating the optimizer
    optimizer = keras.optimizers.Adam(learning_rate=self.lr)

    # Compiling the model
    self.model.compile(loss=keras.losses.MeanSquaredError(), optimizer=optimizer, metrics =['mean_absolute_error'])


    history = self.model.fit(
                self.X_train,
                self.Y_train,
                epochs=self.epoch,
                batch_size=self.batch_size,
                validation_data=(self.X_val, self.Y_val),
                shuffle=False,
                callbacks=[self.trainCallback()]
            )

    return history


  def plot(self,history):
    plt.figure(figsize=(5,5))

    epochs =[idx+1 for idx in range(len(history.history['loss']))]

    plt.plot(epochs, history.history['loss'], label ='loss')
    plt.plot(epochs, history.history['val_loss'], label ='val_loss')
    plt.xlabel('epochs')
    plt.ylabel('loss')
    plt.legend()
    plt.show()

  def predict(self, X):
    return self.model.predict(X)

  def actual_predicted_power_comparison(self):# should be hourly power consumption comparison
    # Comparing the forecasts with the actual values
    yhat = [x[0] for x in self.model.predict(self.X_val)]
    y = [y[0] for y in self.Y_val]

    self.output_rows_actual =self.df_val.iloc[self.lag:,1:]
    self.output_rows_actual.insert(0,ac_name+'_actual_power', y)
    self.output_rows_pred = self.df_val.iloc[self.lag:,1:]
    self.output_rows_pred.insert(0,ac_name+'_predicted_power', yhat)

    self.unscaled_output_actual = self.minmax_new.inverse_transform(self.output_rows_actual)[:,0]
    self.unscaled_output_pred =self.minmax_new.inverse_transform(self.output_rows_pred)[:,0]

    print(self.unscaled_output_actual[:10])
    print(self.unscaled_output_pred[:10])

    plt.figure(figsize=(6,6))
    plt.plot(self.unscaled_output_actual, label ='actual')
    plt.plot(self.unscaled_output_pred, label ='prediction')
    plt.xlabel('Duration from 21st sept to 30th sept, 2019')
    plt.title('actual vs prediction comparison on test set for'+self.ac_name)
    plt.legend()
    plt.show()

  def save(self):
    path ='/content/gdrive/MyDrive/RNN_ac_model_folders'+'/'+self.ac_name.replace(' ','_')
    if(not os.path.exists(path)):
      os.mkdir(path)

    self.model.save(path+'/'+self.ac_name.replace(' ','_')+'_model.h5')
    joblib.dump(self.minmax_new, path+'/'+self.ac_name.replace(' ','_')+'_minmax_scalar.pkl')











```
